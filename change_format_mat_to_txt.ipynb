```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
import scipy.io
mat = scipy.io.loadmat('2021.05.25_2_Analysis.mat')
```

## find time and coordinates

```python
k = mat['Analysis'][0][0]
```

```python
for i in range(len(k)):
    q = k[i]
    print(i, q.shape, len(q))
```

видимо, время и координаты лежат в переменных 5-8, 20

```python
for i in (5, 7, 6, 20, 8):
    print(i, len(k[i][0]), k[i][0].min(), k[i][0].max(), k[i][0].mean())
    if i == 20:
        plt.figure(figsize=(6, 7))
        plt.plot(k[i][0] * -1, color='pink')
    else: 
        plt.figure(figsize=(6, 3))
    plt.plot(k[i][0], color='pink')
    plt.title(str(i))
    plt.show()
```

итог:

время: 8

ось 1: 5 и 7

ось 2: 6 и 20

предлагаю взять 5 и 6 и делить на 10

```python
plt.plot(k[5][0] / 10, k[6][0] / 10, color='palevioletred', linewidth=0.4)
```
